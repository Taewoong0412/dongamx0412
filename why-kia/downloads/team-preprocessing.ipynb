### info() 로 확인한 특이사항 정리하기

1. 왜 점수(score)에 자료형 종류중 float가 있었나???
야구 점수는 2.5점 같은 게 없는데 왜 ??? **[16번 셀 참조]
원래 값은 정수지만 결측값 NaN이 섞여 있어서 pandas가 float 형태로 처리한 것으로 볼 수 있다.
따라서 나중에 취소 경기 등을 처리해서 결측값이 없어지면
8.0 → 8
2.0 → 2
처럼 추후 결측치를 없애고 정수형으로 정리할 후보가 되는 거다.

2. 78개의 득점이 없는 데이터는 0:0으로 끝난경기일까 아니면 기상이변 등으로 인한 경기가 연기되어 없는 데이터 일까?
  데이터를 확인하지 않고 무조건 없는 데이터니 지워야한다 -> 전처리에서 흔하게 범하는 오류
  데이터를 확인한 후 지우는 과정을 통한 전처리를 해야한다

이런 게 바로 전처리다.

### 컬럼 정리전 유의사항
    - 컬럼명으로 중복되는 데이터를 지닌 컬럼을 찾는다.
    - 중복되는 칼럼을 찾게 되면 해당컬럼에 적힌 데이터를 통해 인식할 수 있는 값이어야한다
      (예를 들어 team.id 가 2010이라면 어떤팀인지 알 수없고, team.abbrev(축약) HAN이면 한화라고 유추할 수 있으나 한으로 시작하는 팀이름이
      많을 수도 있다. team.name과 같이 인식이 명확한 값으로 구성된 칼럼이 존재한다면 해당 칼럼을 살린다!

### 데이터 무결성 검증
